# Prompt (paste verbatim)
Write two Python functions cel_to_far(c) and far_to_cel(f) with type hints and docstrings. Build a CLI that asks the user which direction to convert and validates input. Handle bad entries gracefully. Include unit tests for typical and edge cases like -40.

```python
def cel_to_far(c: float) -> float:
    """Convert Celsius to Fahrenheit."""
    return c * 9/5 + 32

def far_to_cel(f: float) -> float:
    """Convert Fahrenheit to Celsius."""
    return (f - 32) * 5/9

def main():
    try:
        direction = input("Convert to (C/F): ").strip().lower()
        temp = float(input("Enter temperature: "))
        if direction == 'c':
            print(f"{temp}°F = {far_to_cel(temp):.2f}°C")
        elif direction == 'f':
            print(f"{temp}°C = {cel_to_far(temp):.2f}°F")
        else:
            print("Invalid direction.")
    except ValueError:
        print("Invalid temperature input.")

main()

# Tests
assert cel_to_far(0) == 32
assert far_to_cel(32) == 0
assert cel_to_far(-40) == -40
assert far_to_cel(-40) == -40
```

```python
def cel_to_far(c: float) -> float:
    """
    Convert Celsius to Fahrenheit.
    Formula: (C × 9/5) + 32
    """
    return c * 9 / 5 + 32

def far_to_cel(f: float) -> float:
    """
    Convert Fahrenheit to Celsius.
    Formula: (F − 32) × 5/9
    """
    return (f - 32) * 5 / 9

def run_converter() -> None:
    """
    CLI tool for temperature conversion.
    Prompts user for direction and temperature, handles errors gracefully.
    """
    while True:
        direction = input("Convert to (C/F), or Q to quit: ").strip().lower()
        if direction == 'q':
            print("Goodbye!")
            break
        if direction not in ('c', 'f'):
            print("❌ Invalid direction. Enter 'C' or 'F'.")
            continue
        try:
            temp_input = input("Enter temperature: ").strip()
            temp = float(temp_input)
            if direction == 'c':
                converted = far_to_cel(temp)
                print(f"{temp}°F = {converted:.2f}°C")
            else:
                converted = cel_to_far(temp)
                print(f"{temp}°C = {converted:.2f}°F")
        except ValueError:
            print("❌ Invalid temperature. Please enter a number.")

# ✅ Unit tests
if __name__ == "__main__":
    assert cel_to_far(0) == 32
    assert far_to_cel(32) == 0
    assert cel_to_far(-40) == -40
    assert far_to_cel(-40) == -40
    assert round(cel_to_far(100), 2) == 212.00
    assert round(far_to_cel(212), 2) == 100.00
    print("All tests passed.\n")
    run_converter()
```

- try/except ensures bad input doesn’t crash the program
- Loop with quit option improves usability
- Type hints and docstrings clarify intent and usage

# PROMPT (paste verbatim)
Write a Python function  that replaces lowercase letters using a given dictionary (e.g., {'a':'4','e':'3'}). Leave other characters unchanged. Return the translated string. Be efficient — avoid repeated string concatenation.


```python
def translate(text, mapping):
    result = []
    for ch in text:
        if ch in mapping:
            result.append(mapping[ch])
        else:
            result.append(ch)
    return ''.join(result)

# Example usage
leet_map = {'a': '4', 'e': '3', 'l': '1', 'o': '0'}
print(translate("hello world", leet_map))
```

# Improved Code

```python
def translate(text: str, mapping: dict[str, str]) -> str:
    """
    Translate lowercase letters in `text` using the provided `mapping`.
    Other characters (including uppercase) are left unchanged.
    """
    if not isinstance(text, str) or not isinstance(mapping, dict):
        raise TypeError("Expected a string and a dictionary.")

    result = [
        mapping[ch] if ch.islower() and ch in mapping else ch
        for ch in text
    ]
    return ''.join(result)

# ✅ Test cases
if __name__ == "__main__":
    leet_map = {'a': '4', 'e': '3', 'l': '1', 'o': '0'}
    assert translate("hello world", leet_map) == "h3110 w0r1d"
    assert translate("HELLO", leet_map) == "HELLO"  # uppercase unchanged
    assert translate("leet", {'e': '3'}) == "l33t"
    assert translate("", leet_map) == ""
    assert translate("123!@#", leet_map) == "123!@#"
    print("All tests passed.")
```

 # Discussion: String Immutability & Efficiency
 Strings are immutable in Python — every modification creates a new string.
 Repeated concatenation like result += ch creates many intermediate strings → O(n²) time in worst case.
 Using a list to accumulate characters and ''.join() at the end is O(n) and memory-efficient.
 This is a classic Python idiom for building strings efficiently.

 -Lecture Concepts Cited
 String immutability: Strings can’t be changed in-place — each .replace() or + creates a new object.
 Efficient iteration: Use list accumulation + ''.join() to avoid quadratic time.
 Character filtering: .islower() ensures only lowercase letters are transformed.
 Mapping logic: Dictionary lookup is O(1) — ideal for character substitution.

Write a Python function normalize_and_find(text, needle) that trims whitespace, converts to lowercase, and returns the index of the first occurrence of needle. Use a safe search method that returns -1 if not found.


```python
# ChatGPT Code
def normalize_and_find(text, needle):
    text = text.strip().lower()
    needle = needle.lower()
    return text.find(needle)

print(normalize_and_find("   Hello World!  ", "world"))
```

# IMPROVED CODE

```python
def normalize_and_find(text: str, needle: str) -> int:
    """
    Normalize the input text by trimming whitespace and converting to lowercase.
    Then return the index of the first occurrence of needle (also lowercased).
    Returns -1 if needle is not found.
    """
    if not isinstance(text, str) or not isinstance(needle, str):
        raise TypeError("Both arguments must be strings.")

    normalized_text = text.strip().lower()
    normalized_needle = needle.lower()
    return normalized_text.find(normalized_needle)

# ✅ Test cases
if __name__ == "__main__":
    assert normalize_and_find("   Hello World!  ", "world") == 6
    assert normalize_and_find("Python Programming", "gram") == 10
    assert normalize_and_find("   ", "x") == -1
    assert normalize_and_find("CaseInsensitive", "case") == 0
    assert normalize_and_find("Mismatch", "match") == 3  # ✅ corrected
    print("All tests passed.")
```

 # Discussion: .find() vs .index()
- .find() returns -1 if the substring is not found — safe and avoids exceptions.
- .index() raises ValueError, which requires a try/except block if used.
- For user-facing or CLI tools, .find() is preferred for graceful failure.

 Lecture Concepts Cited
- String immutability: .strip() and .lower() return new strings, not modify in-place.
- Safe search: .find() avoids exceptions, unlike .index().
- Normalization: Trimming and case conversion are standard preprocessing steps.

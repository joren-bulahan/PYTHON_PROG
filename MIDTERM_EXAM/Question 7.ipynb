# Prompt (paste verbatim)
Write Python classes Vehicle, Car, and Truck. Each has color, mileage, and fuel_liters. Add drive(km, km_per_liter) and refuel(liters). Truck should override drive with different efficiency. Include a test script.

```python
class Vehicle:
    def __init__(self, color, mileage=0, fuel_liters=0):
        self.color = color
        self.mileage = mileage
        self.fuel_liters = fuel_liters

    def drive(self, km, km_per_liter):
        fuel_needed = km / km_per_liter
        if fuel_needed > self.fuel_liters:
            km = self.fuel_liters * km_per_liter
            fuel_needed = self.fuel_liters
        self.mileage += km
        self.fuel_liters -= fuel_needed

    def refuel(self, liters):
        self.fuel_liters += liters

class Car(Vehicle):
    pass

class Truck(Vehicle):
    def drive(self, km, km_per_liter):
        # Trucks are less efficient: reduce efficiency by 20%
        adjusted_efficiency = km_per_liter * 0.8
        super().drive(km, adjusted_efficiency)

# Test script
car = Car("red", fuel_liters=10)
car.drive(50, 10)
print(car.mileage, car.fuel_liters)

truck = Truck("blue", fuel_liters=10)
truck.drive(50, 10)
print(truck.mileage, truck.fuel_liters)
```

# Improved Code

```python
class Vehicle:
    """
    Base class for vehicles with color, mileage, and fuel tracking.
    """
    def __init__(self, color: str, mileage: float = 0.0, fuel_liters: float = 0.0):
        self.color = color
        self.mileage = mileage
        self.fuel_liters = fuel_liters

    def drive(self, km: float, km_per_liter: float) -> None:
        if km_per_liter <= 0:
            raise ValueError("Efficiency must be positive.")
        fuel_needed = km / km_per_liter
        if fuel_needed > self.fuel_liters:
            km = self.fuel_liters * km_per_liter
            fuel_needed = self.fuel_liters
        self.mileage += km
        self.fuel_liters -= fuel_needed

    def refuel(self, liters: float) -> None:
        if liters < 0:
            raise ValueError("Cannot refuel negative liters.")
        self.fuel_liters += liters

class Car(Vehicle):
    """Car inherits Vehicle without changes."""
    pass

class Truck(Vehicle):
    """
    Truck overrides drive with reduced efficiency (20% less).
    """
    def drive(self, km: float, km_per_liter: float) -> None:
        adjusted_efficiency = km_per_liter * 0.8
        super().drive(km, adjusted_efficiency)

# ✅ Test script
if __name__ == "__main__":
    car = Car("red", fuel_liters=10)
    car.drive(50, 10)
    print(f"Car: mileage={car.mileage}, fuel={car.fuel_liters:.2f}")

    truck = Truck("blue", fuel_liters=10)
    truck.drive(50, 10)
    print(f"Truck: mileage={truck.mileage}, fuel={truck.fuel_liters:.2f}")
```

# Discussion: Inheritance & Override
- Car inherits all behavior from Vehicle — no changes needed.
- Truck overrides drive() to reduce efficiency — shows polymorphism.
- super().drive() reuses base logic with modified parameters.

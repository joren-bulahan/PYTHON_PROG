# IMPROVED CODE

```python
def fmt_money(x: float) -> str:
    """
    Format a float as currency with commas and two decimal places.
    Uses ties-to-even rounding (banker's rounding).
    """
    if not isinstance(x, (float, int)):
        raise TypeError("Input must be a float or int.")
    return f"{round(x, 2):,.2f}"

# ✅ Test cases
if __name__ == "__main__":
    assert fmt_money(1234.5) == "1,234.50"
    assert fmt_money(2.5) == "2.50"   # ties to even → 2
    assert fmt_money(3.5) == "4.00"   # ties to even → 4
    assert fmt_money(1234567.891) == "1,234,567.89"
    assert fmt_money(0.005) == "0.00" # rounds down
    assert fmt_money(0.015) == "0.02" # rounds up
    print("All tests passed.")
```

 # Discussion:
    Floating-Point & Ties-to-Even
 	-Python uses IEEE 754 floating-point arithmetic, which defaults to round-to-nearest, ties-to-even — also called banker’s rounding.
 	This avoids rounding bias in large datasets — e.g., round(2.5) →2 , round(3.5) →4 .
 	-Floating-point numbers like 0.1  or 0.005  may not be stored exactly due to binary representation limits — formatting with round()  and f"{.....2f}" helps mitigate display errors.

 # Lecture Concepts Cited

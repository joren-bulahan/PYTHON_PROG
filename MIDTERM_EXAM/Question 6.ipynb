Write a Python function roll() that simulates a fair die (1–6). Then simulate 100,000 rolls of two dice and estimate the probability that their sum equals 7. Compare to the exact probability. Mention random seeding and reproducibility.


```python
import random

def roll():
    return random.randint(1, 6)

def simulate():
    count = 0
    trials = 100000
    for _ in range(trials):
        if roll() + roll() == 7:
            count += 1
    probability = count / trials
    print(f"Estimated probability of sum=7: {probability:.4f}")

simulate()
```

# Improved Code

```python
import random

def roll() -> int:
    """Simulate a fair six-sided die roll."""
    return random.randint(1, 6)

def simulate_dice_sum(target_sum: int = 7, trials: int = 100_000, seed: int = 42) -> float:
    """
    Simulate rolling two dice `trials` times and estimate the probability
    that their sum equals `target_sum`. Uses a fixed seed for reproducibility.
    """
    random.seed(seed)
    count = sum(1 for _ in range(trials) if roll() + roll() == target_sum)
    return count / trials

# ✅ Run simulation
if __name__ == "__main__":
    estimated = simulate_dice_sum()
    exact = 6 / 36  # There are 6 combinations that sum to 7 out of 36 total
    print(f"Estimated probability of sum=7: {estimated:.4f}")
    print(f"Exact probability: {exact:.4f}")
```

# Discussion: Probability of Sum = 7
- Total outcomes for two dice: 6 \times 6 = 36
- Favorable outcomes for sum = 7: (1,6), (2,5), (3,4), (4,3), (5,2), (6,1) → 6 outcomes
- Exact probability: 6 / 36 = 0.1667
# Random Seeding
- random.seed(42) ensures reproducibility — same results every run
- Useful for debugging, testing, and consistent grading

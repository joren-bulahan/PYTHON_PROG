# PROMPT (paste verbatim)
Write a Python function check_password(pw) that checks if a password meets these rules: length ≥ 10, has uppercase, lowercase, digit, symbol (!@#$%^&*), and no spaces. Return a dict of booleans for each rule and overall pass/fail. Add a loop that keeps asking until a valid password is entered or user quits with Q/q.


```python
def check_password(pw):
    rules = {
        "length": len(pw) >= 10,
        "uppercase": any(c.isupper() for c in pw),
        "lowercase": any(c.islower() for c in pw),
        "digit": any(c.isdigit() for c in pw),
        "symbol": any(c in "!@#$%^&*" for c in pw),
        "no_space": " " not in pw
    }
    rules["pass"] = all(rules.values())
    return rules

while True:
    pw = input("Enter password (or Q to quit): ")
    if pw.lower() == "q":
        break
    result = check_password(pw)
    print(result)
    if result["pass"]:
        print("✅ Password accepted.")
        break
    else:
        print("❌ Password failed. Try again.")
```

# Improved Code


```python
def check_password(pw: str) -> dict:
    """
    Validate password against policy:
    - Length ≥ 10
    - At least one uppercase, lowercase, digit, and symbol (!@#$%^&*)
    - No spaces
    Returns a dict of rule results and overall pass/fail.
    """
    rules = {
        "length": len(pw) >= 10,
        "uppercase": any(c.isupper() for c in pw),
        "lowercase": any(c.islower() for c in pw),
        "digit": any(c.isdigit() for c in pw),
        "symbol": any(c in "!@#$%^&*" for c in pw),
        "no_space": " " not in pw
    }
    rules["pass"] = all(rules.values())
    return rules

def prompt_password() -> None:
    """Prompt user until a valid password is entered or they quit."""
    while True:
        pw = input("Enter password (or Q to quit): ").strip()
        if pw.lower() == "q":
            print("Goodbye!")
            break

        result = check_password(pw)
        for rule, passed in result.items():
            if rule != "pass":
                print(f"{rule}: {'✅' if passed else '❌'}")
        if result["pass"]:
            print("✅ Password accepted.")
            break
        else:
            print("❌ Password failed. Try again.\n")

# ✅ Run the tool
if __name__ == "__main__":
    prompt_password()
```

# Discussion: Rule Breakdown
- any(c.isupper() for c in pw) → scans for uppercase letters
- c in "!@#$%^&*" → symbol check via membership test
- " " not in pw → space exclusion
- all(rules.values()) → combines all checks for final verdict

# Improved Code

```python
def exponent_tool() -> None:
    """
    CLI tool that reads two numeric inputs and prints a ** b.
    Handles invalid input gracefully using try/except.
    """
    while True:
        try:
            a_input = input("Enter base (a), or Q to quit: ").strip()
            if a_input.lower() == 'q':
                print("Goodbye!")
                break
            b_input = input("Enter exponent (b): ").strip()
            if b_input.lower() == 'q':
                print("Goodbye!")
                break

            a = float(a_input)
            b = float(b_input)
            result = a ** b
            print(f"{a} ** {b} = {result}")
        except ValueError:
            print("❌ Invalid input. Please enter valid numbers (e.g., 3, -2.5).")

# ✅ Run the tool
if __name__ == "__main__":
    exponent_tool()
```

Discussion: str.isdigit() vs try/except
 - str.isdigit() only returns True for strings of digits — e.g., "42" → True, but "3.5", "-2" → False.
 - It fails for valid numeric formats like floats, negatives, or scientific notation.
 - try/except with float() is more robust — it handles all valid numeric strings and catches errors cleanly.

 Lecture Concepts Cited
 - Exception handling: try/except is the standard for input validation.
 - Numeric parsing: float() supports a wide range of numeric formats.
 - Control flow: Loop with break allows retry and graceful exit.
 - CLI input: input() + .strip() ensures clean user interaction.

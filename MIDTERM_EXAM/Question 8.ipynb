 # Prompt (paste verbatim)
Write Python classes Dog and Kennel. Dog has name, age, breed, and bark(). Kennel holds multiple Dog instances and supports add_dog, remove_dog, oldest_dog(), and find_by_breed(breed). Use composition, not inheritance. Include a test script.


```python
class Dog:
    def __init__(self, name, age, breed):
        self.name = name
        self.age = age
        self.breed = breed

    def bark(self):
        return f"{self.name} says woof!"

class Kennel:
    def __init__(self):
        self.dogs = []

    def add_dog(self, dog):
        self.dogs.append(dog)

    def remove_dog(self, name):
        self.dogs = [dog for dog in self.dogs if dog.name != name]

    def oldest_dog(self):
        return max(self.dogs, key=lambda d: d.age, default=None)

    def find_by_breed(self, breed):
        return [dog for dog in self.dogs if dog.breed == breed]

# Test script
k = Kennel()
k.add_dog(Dog("Buddy", 5, "Labrador"))
k.add_dog(Dog("Max", 7, "Beagle"))
k.add_dog(Dog("Bella", 3, "Beagle"))

print(k.oldest_dog().name)
print([d.name for d in k.find_by_breed("Beagle")])
```

# Improved Code


```python
class Dog:
    """
    Represents a dog with name, age, and breed.
    """
    def __init__(self, name: str, age: int, breed: str):
        self.name = name
        self.age = age
        self.breed = breed

    def bark(self) -> str:
        return f"{self.name} says woof!"

class Kennel:
    """
    Manages a collection of Dog instances.
    """
    def __init__(self):
        self.dogs: list[Dog] = []

    def add_dog(self, dog: Dog) -> None:
        if not isinstance(dog, Dog):
            raise TypeError("Only Dog instances can be added.")
        self.dogs.append(dog)

    def remove_dog(self, name: str) -> None:
        self.dogs = [dog for dog in self.dogs if dog.name != name]

    def oldest_dog(self) -> Dog | None:
        return max(self.dogs, key=lambda d: d.age, default=None)

    def find_by_breed(self, breed: str) -> list[Dog]:
        return [dog for dog in self.dogs if dog.breed == breed]

# ✅ Test script
if __name__ == "__main__":
    kennel = Kennel()
    kennel.add_dog(Dog("Buddy", 5, "Labrador"))
    kennel.add_dog(Dog("Max", 7, "Beagle"))
    kennel.add_dog(Dog("Bella", 3, "Beagle"))

    assert kennel.oldest_dog().name == "Max"
    assert [d.name for d in kennel.find_by_breed("Beagle")] == ["Max", "Bella"]
    print("All tests passed.")
```

# Discussion: Composition vs. Inheritance
- Composition: Kennel has Dogs — ideal for managing collections.
- Inheritance: Would be wrong here — Kennel is not a type of Dog.
- Composition promotes modularity and reuse without forcing type relationships.

Create a Python package helpers/ with modules string_utils.py and math_utils.py. Each has one function: shout(s) and area(l, w). Then write main.py that imports both using import ... as ... and from ... import .... Explain namespace collisions and why aliases help.

# Improved Code

```python
import helpers.string_utils as su
from helpers.math_utils import area

def main():
    print(su.shout("hello world"))
    print(f"Area: {area(5, 3)}")

if __name__ == "__main__":
    main()
```

# Discussion: Namespace Collisions & Aliases
- If both modules had a function named area, using from ... import area could cause a collision — one would overwrite the other.
- Using import ... as ... (e.g., helpers.string_utils as su) keeps functions namespaced — avoids conflicts and improves clarity.
- Aliases shorten long module paths and make code cleaner.

# Lecture Concepts Cited
- Modular design: Splitting logic into reusable files
- Import styles: import ... as ... vs from ... import ...
- Namespace safety: Avoiding collisions with aliases
